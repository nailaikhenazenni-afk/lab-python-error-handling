````python
# Step 1:
def initialize_inventory(products):
    inventory = {}
    for product in products:
        valid_quantity = False
        while not valid_quantity:
            try:
                quantity = int(input(f"Enter the quantity of {product}s available: "))
                if quantity < 0:
                    raise ValueError("Invalid quantity! Please enter a non-negative value.")
                valid_quantity = True
            except ValueError as error:
                print(f"Error: {error}")
        inventory[product] = quantity
    return inventory
    
````

````python
# Step 2:
def calculate_total_price(customer_orders):
    total_price = 0.0
    
    for product in customer_orders:
        while True:
            entry = input(f"Enter the price of {product}: ")
            try:
                price = float(entry)
                if price < 0:
                    print("Error: Price cannot be negative. Please try again.")
                else:
                    total_price += price
                    break
            except ValueError:
                print(f"Error: '{entry}' is not a valid number. Please try again.")
                
    return total_price
````

````python
# Step 3:
def get_customer_orders(inventory):
    # 1. Validar el número de pedidos (num_orders)
    while True:
        try:
            entry = input("How many products do you want to order? ")
            num_orders = int(entry)
            if num_orders < 0:
                print("Error: The number of orders cannot be negative.")
                continue
            break
        except ValueError:
            print(f"Error: '{entry}' is not a valid number. Please enter an integer.")

    customer_orders = {}

    # 2. Bucle de pedidos con validaciones específicas
    for i in range(num_orders):
        while True:
            # Normalizamos la entrada para coincidir con las llaves de inventory
            product_name = input(f"Enter the name for product {i + 1}: ").lower().strip()
            
            try:
                # Caso A: El producto no existe en el catálogo
                if product_name not in inventory:
                    raise ValueError(f"Product '{product_name}' was not found in our catalog.")
                
                # Caso B: El producto existe pero no tiene stock inicial
                if inventory[product_name] == 0:
                    raise ValueError(f"Sorry, '{product_name}' is currently out of stock.")
                
                # Caso C: Validar contra lo ya pedido en esta sesión
                current_count = customer_orders.get(product_name, 0)
                if current_count >= inventory[product_name]:
                    raise ValueError(f"Cannot add more '{product_name}'. Only {inventory[product_name]} available.")

                # Si pasa todas las aduanas, lo añadimos
                customer_orders[product_name] = current_count + 1
                print(f"-> '{product_name}' added to your order.")
                break 

            except ValueError as e:
                # Imprimimos el mensaje específico que lanzamos arriba
                print(f"Invalid Entry: {e}")

    return customer_orders

    
````

````python
def manage_customer_orders():
    products_list = ['t-shirt', 'mug', 'hat', 'book', 'keychain']
    
    # 1. Inicializar el inventario (se asume que esta función existe)
    inventory = initialize_inventory(products_list)
    
    # 2. Obtener los pedidos del cliente (usando la función que validamos antes)
    customer_orders = get_customer_orders(inventory)

    # 3. Mostrar el resumen de lo pedido
    print(f"\nSummary of your order: {customer_orders}")

    # 4. Actualizar el inventario real restando lo pedido
    for product, quantity in customer_orders.items():
        inventory[product] -= quantity

    # 5. Mostrar el inventario final actualizado
    print("\nFinal Inventory after orders:")
    for product, stock in inventory.items():
        print(f"{product}: {stock}")

    return customer_orders

# --- ÚNICA LÍNEA PARA ARRANCAR TODO EL PROGRAMA ---
manage_customer_orders()
````
