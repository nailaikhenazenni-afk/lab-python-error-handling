````python
# Step 1:
def initialize_inventory(products):
    inventory = {}
    for product in products:
        valid_quantity = False
        while not valid_quantity:
            try:
                quantity = int(input(f"Enter the quantity of {product}s available: "))
                if quantity < 0:
                    raise ValueError("Invalid quantity! Please enter a non-negative value.")
                valid_quantity = True
            except ValueError as error:
                print(f"Error: {error}")
        inventory[product] = quantity
    return inventory
````

````python
def calculate_total_price(customer_orders):
    total_price = 0.0
    
    for product in customer_orders:
        while True:
            entry = input(f"Enter the price of {product}: ")
            try:
                price = float(entry)
                if price < 0:
                    print("Error: Price cannot be negative.")
                else:
                    # Si el precio es válido, lo sumamos y salimos del bucle
                    total_price += price
                    break
            except ValueError:
                print(f"Error: '{entry}' is not a valid number. Please try again.")
                
    return total_price
````

````python
# Step 3:
def get_customer_orders(inventory):
    customer_orders = {}
    for i in range(3):  # Assuming a maximum of 5 products can be ordered
        while True:
            product_name = input(f"Enter the name for product {i + 1}: ").lower().strip()
            
            try:
                # Consistencia: ¿Existe el producto?
                if product_name not in inventory:
                    raise ValueError(f"'{product_name}' does not exist in our catalog.")
                
                # Consistencia: ¿Hay stock físico inicial?
                if inventory[product_name] <= 0:
                    raise ValueError(f"'{product_name}' is completely out of stock.")
                
                # Consistencia: ¿Queda stock considerando lo que ya pedí en esta sesión?
                already_ordered = customer_orders.get(product_name, 0)
                if already_ordered >= inventory[product_name]:
                    raise ValueError(f"No more units of '{product_name}' available.")

                # Si pasa las validaciones, se registra
                customer_orders[product_name] = already_ordered + 1
                break 

            except ValueError as e:
                print(f"Error: {e} Please try again.")

    return customer_orders
````

````python
# step 4:
def manage_customer_orders():
    products_list = ['t-shirt', 'mug', 'hat', 'book', 'keychain']
    
    # 1. Creamos el inventario
    inventory = initialize_inventory(products_list)
    
    # 2. Obtenemos los pedidos (ESTO debe ir antes del precio)
    customer_orders = get_customer_orders(inventory)
    
    # 3. Calculamos el precio basándonos en los pedidos recibidos
    total_price = calculate_total_price(customer_orders)
    
    # 4. MOSTRAMOS los resultados en la terminal
    print(f"\n--- Final Summary ---")
    print(f"Products ordered: {customer_orders}")
    print(f"Total price: ${total_price:.2f}")
    
    return total_price

# --- EL PASO FINAL: Ejecutar la función ---
# Sin esta línea, no aparecerá nada en tu pantalla.
manage_customer_orders()
````
